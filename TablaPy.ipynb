```python
!pip install tabula-py[jpype]
import tabula
import pandas as pd
```

```python
tablas  = tabula.read_pdf("sb2.pdf",pages= "all",multiple_tables=False,)
```

```python
df=tablas[0]
```

```python
print(df.head(10))
```

```python
df['Ganador']=df['Ganador'].str.replace('\r','  ').str.rstrip('(0123456789)')
df['Estadio']=df['Estadio'].str.rstrip('(0123456789)')
df['Perdedor']=df['Perdedor'].str.replace('\r','  ').str.rstrip('(0123456789)')
df['Ciudad']=df['Ciudad'].str.rstrip('(0123456789)').str.replace('\r',' ')
df['Fecha']=df['Fecha'].str.replace('\r',' ')
df['Marcador']=df ['Marcador'].str.replace(r'\[.*\]','',regex=True)
```

```python
df['Fecha'] = df['Fecha'].str.replace(
    r'de(enero|febrero|marzo|abril|mayo|junio|julio|agosto|septiembre|octubre|noviembre|diciembre)de',
    r'de \1 de',
    regex=True
)
```
